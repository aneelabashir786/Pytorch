# Mathematical Operations

```python
x = torch.rand(2,2)
x
```

```python
# add
x+2
# sub
x-2
# mul
x*2
# div
x/2
# int div
(x*100)//3
# mod
((x*100)//3)%2
# power
x**2
```

```python
torch.abs(x)
torch.neg(x)
torch.round(x)
torch.ceil(x)
torch.floor(x)
torch.clamp(x , min = 0.2 , max = 0.8)
# sum of whole matrix
torch.sum(x)
torch.sum(x , dim =0 )
torch.sum(x,dim=1)
torch.mean(x)
torch .median(x)
torch.std(x)
torch.var(x)
torch.argmax(x)
torch.argmin(x)
```
