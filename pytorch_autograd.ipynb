torch.autograd is PyTorch's built-in automatic differentiation engine that calculates the gradients of operations performed on tensors. This functionality is crucial for training neural networks, as it allows for the efficient implementation of the backpropagation algorithm using gradient descent without requiring the user to manually derive complex derivatives.

```python
import torch

x = torch.tensor(6.7)
y = torch.tensor(0.0)

w = torch.tensor(1.0)
b = torch.tensor(0.0)
```

```python
def binary_cross_entropy(y_pred, target):
  epsilon = 1e-8
  y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)
  return -target * torch.log(y_pred) - (1 - target) * torch.log(1 - y_pred)
```

```python
# forward pass

z = w * x + b
y_pred = torch.sigmoid(z)
loss = binary_cross_entropy(y_pred , y)
loss
```

```python
dloss_dy_pred = (y_pred-y)/(y_pred*(1-y_pred))
dy_pred_dz = y_pred*(1-y_pred)

dz_dw = x
dz_db = 1

dl_dw = dloss_dy_pred * dy_pred_dz * dz_dw
dl_db = dloss_dy_pred * dy_pred_dz * dz_db
```

```python
print(f"manual gradient loss wrt to weights : {dl_dw}")
print(f"manual gradient loss wrt to bias : {dl_db}")
```

```python
x = torch.tensor(6.7)
y = torch.tensor(0.0)

w = torch.tensor(1.0 , requires_grad=True)
b = torch.tensor(0.0 , requires_grad=True)

z = w * x + b
y_pred = torch.sigmoid(z)
loss = binary_cross_entropy(y_pred , y)
loss.backward()
```
